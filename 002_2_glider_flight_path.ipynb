```python
altitude_m = 10
distance_m = 0
time_s = 0

forward_speed_mps = 5
vertical_speed_mps = -2
step_seconds = 1

print(time_s, "s | ", distance_m, "m forward | ", altitude_m, "m high")

time_s = time_s + step_seconds
distance_m = distance_m + forward_speed_mps * step_seconds
altitude_m = altitude_m + vertical_speed_mps * step_seconds

print(time_s, "s | ", distance_m, "m forward | ", altitude_m, "m high")
```

```python
altitude_m = 10
distance_m = 0
time_s = 0


forward_speed_mps = 5
vertical_speed_mps = -2
step_seconds = 1


distances_m = [distance_m]
altitudes_m = [altitude_m]
times_s = [time_s]

print(time_s, "s | ", distance_m, "m forward |", altitude_m, "m high")


while altitude_m > 0:

    time_s = time_s + step_seconds
    distance_m = distance_m + forward_speed_mps * step_seconds
    altitude_m = altitude_m + vertical_speed_mps * step_seconds

    distances_m.append(distance_m)
    altitudes_m.append(altitude_m)
    times_s.append(time_s)

    print(time_s, "s | ", distance_m, "m forward |", altitude_m, "m high")
```

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(8, 4))
plt.plot(distances_m, altitudes_m, marker="o")
for x, y, t in zip(distances_m, altitudes_m, times_s):
  plt.annotate(f"{t:g} s", (x,y), xytext=(5,5), textcoords="offset points")
plt.axhline(0, color="black", linewidth=1)


plt.title("Wrightsville Glider - First Flight")
plt.xlabel("Forward distance from launch (meters)")
plt.ylabel("Height above the beach (meters)")
plt.grid(True)
plt.show()
```
